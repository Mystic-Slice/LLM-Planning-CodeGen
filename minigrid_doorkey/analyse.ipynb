```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("metrics_doorkey.csv")
df
```

```python
df.groupby('method_name')['reward'].describe()[['mean', 'std']].round(2)
```

```python
grouped = df.groupby('method_name')['reward'].agg(['mean', 'std', 'count'])
grouped['ci95'] = 1.96 * grouped['std'] / np.sqrt(grouped['count'])
grouped[['mean', 'ci95']].round(2)
```
